```python
import json
import os
```

```python
for idea_json in ['ai_scientist_baseline_ideas.json', 'ai_scientist_persona_baseline_ideas.json'] :
    with open(os.path.join('templates/ppo_with_folds', idea_json)) as f:
        ideas = json.load(f)
    for idea in ideas:
        if idea["Name"] not in [ideas[0]["Name"], ideas[1]["Name"]]:
            try : 
                with open(os.path.join('results/ppo_with_folds', idea["Name"], f"{idea['Name']}_aider.txt"), "r") as f:
                    text = f.read()
                num_instances = text.count(f"python experiment.py --out_dir={idea['Name']}")
                if num_instances < 2 : 
                    print("Didn't finishing making changes", idea["Name"], idea_json)
                with open(os.path.join('results/ppo_with_folds', idea["Name"], idea["Name"], "final_info.json"), "r") as f:
                    final_info = json.load(f)
                if final_info["reward"] == -12.711948504884596 :
                    print("No changes made", idea["Name"], idea_json)  
            except FileNotFoundError:
                print("File not found", idea["Name"], idea_json)
                continue
```
